```python
# Gaussian eliminaation method (forward eliminaiton  + back substitution)
# solves :
# 2x + y - z = 8
#-3x + y + 2z = -11
#-2x + y + 2z = -3
import numpy as np

# augmented matrix [A | b]
A = np.array([
    [ 2.,  1., -1.,   8.],
    [-3., -1.,  2., -11.],
    [-2.,  1.,  2.,  -3.]
])

n = len(A)

# step 1 : forward elimination
# make everything below each pivot = 0
for i in range(n):
    pivot = A[i][i] # current pivot element
    # eliminate entries below the pivot
    for j in range(i+1, n):
        factor = A[j][i] / pivot
        A[j] = A[j] - factor * A[i]
print("Upper Triangular Matrix:")
print(A)

# step 2 : back substitution
# solve from bottom row upward
x = np.zeros(n)
for i in range(n-1, -1, -1):
    x[i] = (A[i][-1] - np.dot(A[i]
[i+1:n], x[i+1:n])) / A[i][i]

print("\nSolution:")
print(f"x = {x[0]:.2f}")
print(f"y = {x[1]:.2f}")
print(f"z = {x[2]:.2f}")
```

```python
# 2x + y = 5
# x + 3y = 6
import numpy as np
A = np.array([[2, 1], [1, 3]]) # coefficcient matrix
b = np.array([5, 6 ]) # right hand side
x = np.linalg.solve(A, b)
# numpy uses optimised linear algebra internally
print(x)
```

```python
# 2x + y -z = 8
#-3x - y +2z =-11
#-2x + y + 2z = -3
import numpy as np
# coefficient matrix
A = np.array([
    [2,  1, -1],
    [-3, -1, 2],
    [-2,  1, 2]
], dtype=float)
# right hand side
b = np.array([8, -11, -3], dtype=float)
# solve the system of equations
x = np.linalg.solve(A, b)
print("x =", x[0])
print("y =", x[1])
print("z =", x[2])
```
