```python
# Gaussian eliminaation method (forward eliminaiton  + back substitution)
# solves :
# 2x + y - z = 8
#-3x + y + 2z = -11
#-2x + y + 2z = -3
import numpy as np

# augmented matrix [A | b]
A = np.array([
    [ 2.,  1., -1.,   8.],
    [-3., -1.,  2., -11.],
    [-2.,  1.,  2.,  -3.]
])

n = len(A)

# step 1 : forward elimination
# make everything below each pivot = 0
for i in range(n):
    pivot = A[i][i] # current pivot element
    # eliminate entries below the pivot
    for j in range(i+1, n):
        factor = A[j][i] / pivot
        A[j] = A[j] - factor * A[i]
print("Upper Triangular Matrix:")
print(A)

# step 2 : back substitution
# solve from bottom row upward
x = np.zeros(n)
for i in range(n-1, -1, -1):
    x[i] = (A[i][-1] - np.dot(A[i]
[i+1:n], x[i+1:n])) / A[i][i]

print("\nSolution:")
print(f"x = {x[0]:.2f}")
print(f"y = {x[1]:.2f}")
print(f"z = {x[2]:.2f}")
```

```python
# ============================================================
# LINEAR REGRESSION FROM SCRATCH USING GRADIENT DESCENT
# ============================================================
# What we're building:
# 1. Generate synthetic data with a known linear relationship
# 2. Implement Linear Regression using Gradient Descent (no sklearn!)
# 3. Visualize the cost function in 3D to see how gradient descent works
# 4. Watch the algorithm converge to the optimal solution

print("="*70)
print("LINEAR REGRESSION FROM SCRATCH WITH GRADIENT DESCENT")
print("="*70)
print("\n🎯 GOAL: Find the best line y = mx + b that fits our data")
print("📈 METHOD: Gradient Descent - iterative optimization")
print("🎨 BONUS: 3D visualization of the cost function surface")
print("="*70)

# Step 1: Import libraries
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
from matplotlib import cm
import warnings
warnings.filterwarnings('ignore')

print("\n✓ Libraries imported successfully!")

# ============================================================
# STEP 2: Generate Synthetic Data
# ============================================================
print("\n" + "="*70)
print("STEP 1: GENERATING SYNTHETIC DATA")
print("="*70)

# Set random seed for reproducibility
np.random.seed(42)

# Generate 100 data points
n_samples = 100
X = np.random.randn(n_samples, 1) * 10  # Random X values between -20 and 20

# True relationship: y = 2.5x + 5 + noise
true_slope = 2.5
true_intercept = 5.0
noise = np.random.randn(n_samples, 1) * 3  # Add some randomness
y = true_slope * X + true_intercept + noise

# Flatten for easier manipulation
X_flat = X.flatten()
y_flat = y.flatten()

print(f"📊 Generated {n_samples} data points")
print(f"📐 True relationship: y = {true_slope}x + {true_intercept}")
print(f"🔊 Added Gaussian noise with std=3")

# Visualize the data
plt.figure(figsize=(10, 6))
plt.scatter(X_flat, y_flat, alpha=0.6, label='Data points')
plt.xlabel('X (Feature)')
plt.ylabel('y (Target)')
plt.title('Generated Data with Linear Relationship')
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()

print("\n✓ Data generated and visualized!")

# ============================================================
# STEP 3: Implement Linear Regression using Gradient Descent
# ============================================================
print("\n" + "="*70)
print("STEP 2: IMPLEMENTING GRADIENT DESCENT")
print("="*70)

class LinearRegressionGD:
    """
    Linear Regression using Gradient Descent from scratch
    """
    def __init__(self, learning_rate=0.01, n_iterations=1000):
        """
        Parameters:
        - learning_rate: Step size for gradient descent
        - n_iterations: Number of iterations to run
        """
        self.learning_rate = learning_rate
        self.n_iterations = n_iterations
        self.weights = None  # Slope (m)
        self.bias = None     # Intercept (b)
        self.cost_history = []  # Track cost for visualization

    def compute_cost(self, X, y):
        """
        Compute Mean Squared Error (MSE) cost
        Cost = (1/n) * Σ(y_pred - y_actual)²
        """
        n_samples = len(X)
        y_pred = self.predict(X)
        cost = (1/n_samples) * np.sum((y_pred - y) ** 2)
        return cost

    def predict(self, X):
        """
        Make predictions: y = mx + b
        """
        return self.weights * X + self.bias

    def fit(self, X, y):
        """
        Train the model using Gradient Descent
        """
        n_samples = len(X)

        # Initialize parameters to 0
        self.weights = 0.0
        self.bias = 0.0
        self.cost_history = []

        print("🔄 Training started...")
        print(f"   Learning rate: {self.learning_rate}")
        print(f"   Iterations: {self.n_iterations}")
        print("-" * 50)

        for iteration in range(self.n_iterations):
            # Forward pass: make predictions
            y_pred = self.predict(X)

            # Calculate gradients
            # Gradient of weights: dw = (2/n) * Σ(y_pred - y) * X
            dw = (2/n_samples) * np.sum((y_pred - y) * X)
            # Gradient of bias: db = (2/n) * Σ(y_pred - y)
            db = (2/n_samples) * np.sum(y_pred - y)

            # Update parameters (move opposite to gradient)
            self.weights -= self.learning_rate * dw
            self.bias -= self.learning_rate * db

            # Store cost for visualization
            cost = self.compute_cost(X, y)
            self.cost_history.append(cost)

            # Print progress every 100 iterations
            if iteration % 100 == 0:
                print(f"   Iteration {iteration:4d}: Cost = {cost:.4f}, m = {self.weights:.4f}, b = {self.bias:.4f}")

        print("-" * 50)
        print("✅ Training complete!")
        print(f"   Final parameters:")
        print(f"   - Slope (m): {self.weights:.4f}")
        print(f"   - Intercept (b): {self.bias:.4f}")
        print(f"   - Final cost: {self.cost_history[-1]:.4f}")
        print(f"   - True values: m={true_slope}, b={true_intercept}")

# ============================================================
# STEP 4: Train the Model
# ============================================================
print("\n" + "="*70)
print("STEP 3: TRAINING THE MODEL")
print("="*70)

# Create and train the model
model = LinearRegressionGD(learning_rate=0.001, n_iterations=1000)
model.fit(X_flat, y_flat)

# ============================================================
# STEP 5: Visualize Results
# ============================================================
print("\n" + "="*70)
print("STEP 4: VISUALIZING RESULTS")
print("="*70)

# 5.1: Plot the fitted line vs data
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.scatter(X_flat, y_flat, alpha=0.6, label='Data points')

# Plot the true line
X_line = np.linspace(X_flat.min(), X_flat.max(), 100)
y_true_line = true_slope * X_line + true_intercept
plt.plot(X_line, y_true_line, 'g-', label='True line', linewidth=2)

# Plot the fitted line
y_pred_line = model.predict(X_line)
plt.plot(X_line, y_pred_line, 'r--', label='Fitted line', linewidth=2)

plt.xlabel('X')
plt.ylabel('y')
plt.title('Linear Regression Results')
plt.legend()
plt.grid(True, alpha=0.3)

# 5.2: Plot the cost function convergence
plt.subplot(1, 2, 2)
plt.plot(model.cost_history, 'b-', linewidth=2)
plt.xlabel('Iteration')
plt.ylabel('Cost (MSE)')
plt.title('Cost Function Convergence')
plt.grid(True, alpha=0.3)
plt.yscale('log')  # Log scale to see the convergence better

plt.tight_layout()
plt.show()

print("\n✓ Results visualized!")
print("📊 Left: Data with true line (green) and fitted line (red dashed)")
print("📉 Right: Cost function decreasing over iterations")

# ============================================================
# STEP 6: 3D Visualization of Cost Function
# ============================================================
print("\n" + "="*70)
print("STEP 5: 3D VISUALIZATION OF COST FUNCTION")
print("="*70)
print("🎨 This shows how gradient descent finds the minimum")
print("   on the cost surface!")

# Create a grid of m and b values
m_vals = np.linspace(0, 5, 50)
b_vals = np.linspace(0, 10, 50)
M, B = np.meshgrid(m_vals, b_vals)

# Calculate cost for each combination
Z = np.zeros_like(M)
n_samples = len(X_flat)

for i in range(len(m_vals)):
    for j in range(len(b_vals)):
        m = m_vals[i]
        b = b_vals[j]
        y_pred = m * X_flat + b
        Z[j, i] = (1/n_samples) * np.sum((y_pred - y_flat) ** 2)

# Create 3D plot
fig = plt.figure(figsize=(14, 6))

# Plot 1: Surface plot
ax1 = fig.add_subplot(1, 2, 1, projection='3d')
surf = ax1.plot_surface(M, B, Z, cmap=cm.viridis, alpha=0.8,
                        linewidth=0, antialiased=True)

# Plot the path of gradient descent
# Extract the parameters at each iteration (sample every 10 steps)
path_m = []
path_b = []
path_cost = []

# Re-run gradient descent to get path
m_temp, b_temp = 0.0, 0.0
learning_rate = 0.001
for i in range(0, 1000, 10):  # Sample every 10 iterations
    y_pred = m_temp * X_flat + b_temp
    dw = (2/n_samples) * np.sum((y_pred - y_flat) * X_flat)
    db = (2/n_samples) * np.sum(y_pred - y_flat)
    m_temp -= learning_rate * dw
    b_temp -= learning_rate * db

    path_m.append(m_temp)
    path_b.append(b_temp)
    cost = (1/n_samples) * np.sum((m_temp * X_flat + b_temp - y_flat) ** 2)
    path_cost.append(cost)

# Plot the path
ax1.scatter(path_m, path_b, path_cost, c='red', s=20,
            label='Gradient Descent Path', alpha=0.8)
ax1.scatter([model.weights], [model.bias], [model.cost_history[-1]],
            c='yellow', s=100, marker='*', label='Final Solution',
            edgecolors='black', linewidth=2)

ax1.set_xlabel('Slope (m)')
ax1.set_ylabel('Intercept (b)')
ax1.set_zlabel('Cost (MSE)')
ax1.set_title('Cost Function Surface with Gradient Descent Path')
ax1.legend()

# Plot 2: Contour plot (bird's eye view)
ax2 = fig.add_subplot(1, 2, 2)
contour = ax2.contour(M, B, Z, levels=20, cmap='viridis')
ax2.clabel(contour, inline=True, fontsize=8)

# Plot the path on contour
ax2.plot(path_m, path_b, 'r-', linewidth=2, label='Gradient Descent Path')
ax2.scatter(path_m[0], path_b[0], c='blue', s=100, marker='o',
            label='Start', edgecolors='black')
ax2.scatter(path_m[-1], path_b[-1], c='yellow', s=100, marker='*',
            label='End', edgecolors='black', linewidth=2)
ax2.scatter(true_slope, true_intercept, c='green', s=100, marker='x',
            label='True Solution', linewidth=2)

ax2.set_xlabel('Slope (m)')
ax2.set_ylabel('Intercept (b)')
ax2.set_title('Contour Map of Cost Function')
ax2.legend()
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

print("\n✓ 3D Visualization complete!")
print("🔴 Red dots: Path of gradient descent from start to minimum")
print("⭐ Yellow star: Final solution found by gradient descent")
print("❌ Green X: True solution (target)")
print("📈 Surface height: Cost value (lower is better)")

# ============================================================
# STEP 7: Model Evaluation
# ============================================================
print("\n" + "="*70)
print("STEP 6: MODEL EVALUATION")
print("="*70)

# Calculate R² score
y_pred_final = model.predict(X_flat)
ss_tot = np.sum((y_flat - np.mean(y_flat)) ** 2)
ss_res = np.sum((y_flat - y_pred_final) ** 2)
r2_score = 1 - (ss_res / ss_tot)

# Calculate MSE
mse = model.cost_history[-1]

print("📊 Performance Metrics:")
print(f"   - Mean Squared Error (MSE): {mse:.4f}")
print(f"   - R² Score: {r2_score:.4f}")
print(f"   - True vs Estimated:")
print(f"     • Slope (m): {true_slope} vs {model.weights:.4f} (error: {abs(true_slope - model.weights):.4f})")
print(f"     • Intercept (b): {true_intercept} vs {model.bias:.4f} (error: {abs(true_intercept - model.bias):.4f})")

print("\n💡 R² Interpretation:")
print(f"   • R² = {r2_score:.4f} means {r2_score*100:.1f}% of variance is explained")
if r2_score > 0.9:
    print("   • Excellent fit! Model captures the relationship well.")
elif r2_score > 0.7:
    print("   • Good fit! Most variance explained.")
else:
    print("   • Consider adding more features or complexity.")

# ============================================================
# STEP 8: Making Predictions
# ============================================================
print("\n" + "="*70)
print("STEP 7: MAKING PREDICTIONS")
print("="*70)

# Test with new values
test_X = np.array([-15, -5, 0, 5, 15])
predictions = model.predict(test_X)
true_values = true_slope * test_X + true_intercept

print("📈 Sample Predictions:")
print("   X  |  Predicted y  |  True y  |  Error")
print("-" * 50)
for x, pred, true in zip(test_X, predictions, true_values):
    error = abs(pred - true)
    print(f"  {x:4.0f}  |    {pred:7.2f}    |  {true:6.2f}  |   {error:.2f}")

# ============================================================
# FINAL SUMMARY
# ============================================================
print("\n" + "="*70)
print("SUMMARY: WHAT WE LEARNED TODAY")
print("="*70)
print("""
🎯 WHAT WE BUILT:
   ✓ Linear Regression from scratch using Gradient Descent
   ✓ No sklearn - pure NumPy implementation
   ✓ 3D visualization of the cost function
   ✓ Real-time convergence tracking

📚 KEY CONCEPTS LEARNED:
   1. GRADIENT DESCENT: Algorithm that finds minimum of a function
      - Compute gradients (derivatives) of the cost function
      - Move in direction opposite to the gradient
      - Learning rate controls step size

   2. COST FUNCTION: Mean Squared Error (MSE)
      - Measures how wrong our model is
      - Lower = better fit to data
      - Convex function (only one minimum)

   3. 3D VISUALIZATION:
      - Surface shows all possible (m,b) combinations
      - Gradient descent rolls down to the lowest point
      - Path shows the algorithm's journey

💡 WHY THIS MATTERS:
   - Foundation of ALL machine learning
   - Same principles used in neural networks
   - Understanding gradient descent is crucial for AI
   - Many advanced algorithms use variations of this

🔗 NEXT STEPS:
   - Try with different learning rates
   - Add polynomial features (non-linear regression)
   - Implement stochastic gradient descent
   - Add regularization (Ridge/Lasso regression)
""")

print("="*70)
print("🎉 CONGRATULATIONS! You've implemented Linear Regression from scratch!")
print("   You now understand the core of how AI learns!")
print("="*70)
```
